## Keyword and Semantic Search

To discuss the results, we will be looking at 5 of the 12 queries that we ran: 2 easy, 2 medium, and 1 complex. The rest of the results are also present to allow the reader to get an overall sense of how well these search systems performed over various different queries, in addition to those we discuss.

### Easy Queries Discussion

#### Query: "knitting crochet guide"

Our BM25 search method did quite well with this query. A total of 4/5 of the results were related to either knitting or crocheting - so actually useful. This was to be expected as these are popular keywords that almost certainly would be used in searching for books on these topics. The addition of the word 'guide' must have been why an irrelevant result surfaced - a book about colouring. 

The semantic search also performed well. There were several books that overlapped in both search results. It also returned a book that wasn't related to knitting or crocheting. However, interestingly enough, the book was one on embroidery. In the world of fiber arts, this could be seen to be more similar to knitting and crocheting that colouring would be. So even though not in our search query, we can think of this at least in the same family of crafts. (May be useful for variety - given we haven't exhaused the books on knitting and crocheting in our data sample.)

#### Query: "personal finance debt free"

For this search, we provided several words that BM25 should have had no trouble returning relevant books. And this was the result - every book had some element of taking charge of ones finances. The results are also varied (e.g., retirement, women-focused), so depending on the user, they would either be fine with the results or have to add additional keywords to narrow down the focus.

Semantic search results also returned some of the same books. There were some different ones too, but these were still related to the search and would have been useful.

Overall, from both of these queries we can see that BM25 and the semantic search performed well. Almost every result was useful. These types of easy searches definitely play to keyword strength, however, we didn't see that the semantic search lacked in performance. However, it was noted that if we want a book on the same subject or very close to it, the semantic search would be able to find a closer 'match'. As seen with our embroidery vs. colouring book comparison.

### Medium Queries Discussion

These queries were designed to be slightly harder than the easier ones. They are used to see if semantic search can notably out-perform keyword search.

#### Query "running a store successfully"

BM25 search did not perform well here. Understandably, it can't generalize to what we were looking for beyond one word at a time. None of the books returned would have been useful for this query searcher. However, it did do its job and returned books with "store", "successfully" in the title and book details, but that's about it.

This is were the semantic search shone - every result could contain useful information for the user. Rather than being tied down to only using keywords, by finding similar matched books-to-query, the searcher doesn't have to be explicit in their query. For a topic like this, where the search may be more vague, it really highlights semantic searching's strength. 

#### Query "how to get out of debt"

Opposed to our previous medium query, we included keywords in the way of a phrase that would probably show up in several personal finance books. And BM25 was able to retrieve a couple of books that *may* have been helpful. Not the best results, but something. On the flipside, and not surprisingly, semantic search returned all, potentially, useful books.

For these medium queries, semantic search clearly performed the best out of the two. Not having the constraint of keywords improved the results. We did note, however, that when there's at least one important keyword in these more ambiguous queries, keyword search may be able to get back something useful. Despite this, in these cases, it isn't the search to rely on.

### Complex Query Discussion

#### Query: "how to beat your competitors"

Here we wanted to introduce a query that's a bit more abstract and with keywords that could return several different topics.

Our BM25 results were what we were trying to test. They were books on various different subjects. We got a novel, children's book, American sporting victory, and small business help. Like we showcased in our medium-strength queries, BM25 is not able to capture any nuance and this hinders its ability to return useful books.

The semantic search, of course, did well. They all had something related to winning in life. The results included books on personal development, business, and athletics; this was either through personal growth, business acumen, or the mental mindset of an athlete. Even though these books have different focuses, they can all be seen to tie back to our original search. This isn't semantic searches fault, but rather the user would need to be more specific if, or example, they only wanted business-related books. However, it is good to see that the results didn't just focus on one subject (i.e., only sports competitions). Helps the user see what's out there and to narrow their query or perhaps see a parallel in another field to theirs.
