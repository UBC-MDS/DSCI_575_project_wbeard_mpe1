## Model Selection Reasoning

We decided to use a model through the HuggingFace API Endpoint. It was primarily because this did not require any one running our project to have a GPU. Additionally, as part of other project setup, the user needed to obtain a HuggingFace API token. With that step already completed, it made using the Inference API a nature choice in ensuring accessibility for those for may not have access to a GPU.

The model selected was Meta-Llama-3-8B-Instruct. Having been trained on over 15 trillion tokens of public data, we thought that it would be well suited for generating understandable responses when prompted.


## System Prompt Variants

We started with a prompt that was provided by the lecture notes and tested that to see how it would perform. It performed alright and results were relevant to what a user would be interested in receiving.

Our updates were more on the side of improving and limiting what the response included. We wanted to make sure that only books that had both the title and author available were included. In the cases that there were no relevant results, we wanted to make sure that the response wasn't nonsense and completely unrelated to what the user was asking. This was especially important since our book sample was limited to 10,000 - not every subject/query may have relevant results. And finally, to prevent the user having to read a paragraph and/or the response being cut off after the max tokens has been reached, we made sure to specify that the response be brief.

```
SYSTEM_PROMPT = """
    You are a helpful Amazon shopping assistant.
    You have access to a sample of books listed on Amazon.
    Answer the question using only the following context:
    Always provide the title and author, when available.
    Do not include results that do not have both title and author.
    If there isn't helpful information in the context, just say that
    there are no relevant books available in the books sample.
    Be very brief, answer is less then 40 words.
    """
```

## Overall Thoughts

Our RAG workflow performed really well. Both the RAG semantic and ensemble pipelines returned similar results: understandable and relevant to the search query. What we do want to note is that our book sample is 10,000 books. Depending on how specific the query is, there may simply not have been a large pool of related books for that particular subject. We see that for our first query of "backpacking across Europe" returning ultrarunning results. However, this can also be seen as a benefit because if there aren't many subject-related books, at least what is returned is tangential. 

A couple of limitations we identified are related to the reviews selected and the weightings we applied when creating the ensemble retriever. For each of the books we only selected one top review to create the retriever objects. This could skew the results to focus only on those that are positive. And only one review at that. An improvement could be to include more reviews and not only include the highly rated ones, but randomly sampled from 0 to 5. 

For the ensemble retriever we decided on a weighting of 0.8 coming from the semantic retriever and 0.2 from the BM25. We selected this because our semantic results were better when the query was more abstract and outperformed BM25. However, a potential improvement could be to change these weights to be more in favour of BM25 to capture those books/reviews that may be more heavy on relevant keywords. This isn't what we saw, but with a much larger book sample, this could something worth trying to get improved results.
